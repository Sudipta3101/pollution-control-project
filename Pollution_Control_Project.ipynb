## Objective
Clean and transform pollution monitoring data to improve data quality and generate meaningful insights.

## Technologies Used
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn

#### Importing the Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("pollution_raw_data.csv", encoding='latin1')

# Parse overflowed clock times (for example, 24:10 becomes 00:10 on the next day).
ts_parts = df['TS'].astype('string').str.extract(r'^\s*(\d{2}-\d{2}-\d{4})\s+(\d{1,2}):(\d{2})\s*$')
ts_date = pd.to_datetime(ts_parts[0], format='%d-%m-%Y', errors='coerce')
ts_hours = pd.to_numeric(ts_parts[1], errors='coerce')
ts_minutes = pd.to_numeric(ts_parts[2], errors='coerce')
df['TS_normalized'] = ts_date + pd.to_timedelta(ts_hours, unit='h') + pd.to_timedelta(ts_minutes, unit='m')
```

```python
print("Before Datetime Normalization:")
print(df["TS"].head())

ts = df["TS"].astype(str).str.strip()

mask_24 = ts.str.contains(" 24:", na=False)

ts = ts.str.replace(" 24:", " 00:", regex=False)

df["TS"] = pd.to_datetime(
    ts,
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

df.loc[mask_24, "TS"] = df.loc[mask_24, "TS"] + pd.Timedelta(days=1)

print("\nAfter Datetime Normalization:")
print(df["TS"].head())

print("\nMissing datetime values:", df["TS"].isna().sum())
```

```python
print("Before Standardization:")
print(df["Unit"].unique())

df["Unit"] = (
    df["Unit"].astype(str).str.strip().replace({
        "Âµg/mÂ³": "ug/m3",
        "UG/M3": "ug/m3",
        "mg/nm3": "mg/Nm3"
    })
)

print("\nAfter Standardization:")
print(df["Unit"].unique())

print("\nUnit Counts:")
print(df["Unit"].value_counts())
```

```python
print("Before Parsing:")
print(df["LatLon"].head())

df[["Latitude", "Longitude"]] = df["LatLon"].str.split(",", expand=True)

df["Latitude"] = df["Latitude"].str.strip()
df["Longitude"] = df["Longitude"].str.strip()

df["Latitude"] = pd.to_numeric(df["Latitude"], errors="coerce")
df["Longitude"] = pd.to_numeric(df["Longitude"], errors="coerce")

df.loc[(df["Latitude"] < -90) | (df["Latitude"] > 90), "Latitude"] = pd.NA
df.loc[(df["Longitude"] < -180) | (df["Longitude"] > 180), "Longitude"] = pd.NA

print("\nAfter Parsing:")
print(df[["Latitude", "Longitude"]].head())

print("\nMissing Latitude:", df["Latitude"].isnull().sum())
print("Missing Longitude:", df["Longitude"].isnull().sum())
```

```python
print("Negative PM2.5 values:", (df["PM2_5"] < 0).sum())
print("Impossible SO2 values (99999):", (df["SO2"] == 99999).sum())
print("NOx values equal to 0:", (df["NOx"] == 0).sum())

df.loc[df["PM2_5"] < 0, "PM2_5"] = pd.NA

df.loc[df["SO2"] == 99999, "SO2"] = pd.NA

df.loc[df["NOx"] == 0, "NOx"] = pd.NA

print("\nAfter Cleaning:")
print("Negative PM2.5 values:", (df["PM2_5"] < 0).sum())
print("Impossible SO2 values (99999):", (df["SO2"] == 99999).sum())
print("NOx values equal to 0:", (df["NOx"] == 0).sum())

print("\nMissing Values:")
print(df[["PM2_5", "SO2", "NOx"]].isnull().sum())
```

```python
print("Duplicates Before Merging:")
print(df.duplicated(subset=["Plant_ID", "TS"]).sum())

df = df.groupby(["Plant_ID", "TS"], as_index=False).first()

print("\nDuplicates After Merging:")
print(df.duplicated(subset=["Plant_ID", "TS"]).sum())

print("\nDataset Shape:")
print(df.shape)

print("\nSample Data:")
print(df[["Plant_ID", "TS"]].head())
```

```python
print("Before Calibration:")
print(df[["PM2_5", "SO2", "NOx"]].head())

df["PM2_5"] = (df["PM2_5"] * df["Calibration_span_factor"]) + df["Calibration_zero_offset"]

df["SO2"] = (df["SO2"] * df["Calibration_span_factor"]) + df["Calibration_zero_offset"]

df["NOx"] = (df["NOx"] * df["Calibration_span_factor"]) + df["Calibration_zero_offset"]

df["PM2_5"] = df["PM2_5"].round(2)
df["SO2"] = df["SO2"].round(2)
df["NOx"] = df["NOx"].round(2)

print("\nAfter Calibration:")
print(df[["PM2_5", "SO2", "NOx"]].head())
```

```python
print("Before Filtering:")
print(df["PM2_5"].head(10))

rolling_median = df["PM2_5"].rolling(window=5, center=True).median()

mad = (df["PM2_5"] - rolling_median).abs().rolling(window=5, center=True).median()

threshold = 3 * 1.4826 * mad

outliers = (df["PM2_5"] - rolling_median).abs() > threshold

df.loc[outliers, "PM2_5"] = rolling_median[outliers]

print("\nAfter Filtering:")
print(df["PM2_5"].head(10))

print("\nOutliers Corrected:", outliers.sum())
```

```python
print("Before Gap Filling:")
print(df[["PM2_5", "SO2", "NOx"]].isnull().sum())

df["PM2_5"] = df["PM2_5"].fillna(df["PM2_5"].mean())
df["SO2"] = df["SO2"].fillna(df["SO2"].mean())
df["NOx"] = df["NOx"].fillna(df["NOx"].mean())

print("\nAfter Gap Filling:")
print(df[["PM2_5", "SO2", "NOx"]].isnull().sum())

print("\nSample Data:")
print(df[["PM2_5", "SO2", "NOx"]].head())
```

```python
print("Downtime Labels Before:")
print(df["Status"].value_counts())

df["Downtime"] = "No"

mask = (
    (df["Maintenance_start_ts"].notna()) &
    (df["Maintenance_end_ts"].notna()) &
    (df["TS"] >= df["Maintenance_start_ts"]) &
    (df["TS"] <= df["Maintenance_end_ts"])
)

df.loc[mask, "Downtime"] = "Yes"

print("\nDowntime Labels After:")
print(df["Downtime"].value_counts())

print("\nSample Data:")
print(df[["TS", "Maintenance_start_ts", "Maintenance_end_ts", "Downtime"]].head(10))

```

```python
print("Values Below LOD:")

print("PM2.5:", (df["PM2_5"] < df["LOD_PM2_5"]).sum())
print("SO2:", (df["SO2"] < df["LOD_SO2"]).sum())
print("NOx:", (df["NOx"] < df["LOD_NOx"]).sum())

df.loc[df["PM2_5"] < df["LOD_PM2_5"], "PM2_5"] = df["LOD_PM2_5"] / 2

df.loc[df["SO2"] < df["LOD_SO2"], "SO2"] = df["LOD_SO2"] / 2

df.loc[df["NOx"] < df["LOD_NOx"], "NOx"] = df["LOD_NOx"] / 2

print("\nAfter LOD Handling:")

print("PM2.5:", (df["PM2_5"] < df["LOD_PM2_5"]).sum())
print("SO2:", (df["SO2"] < df["LOD_SO2"]).sum())
print("NOx:", (df["NOx"] < df["LOD_NOx"]).sum())

print("\nSample Data:")
print(df[["PM2_5", "LOD_PM2_5", "SO2", "LOD_SO2", "NOx", "LOD_NOx"]].head())

```

```python
print(df[df["PM2_5"] == df["LOD_PM2_5"] / 2][["PM2_5", "LOD_PM2_5"]].head())
```

```python
print("Before Normalization:")
print(df["Status"].unique())

df["Status"] = df["Status"].str.strip().str.upper()

df["Status"] = df["Status"].replace({
    "MAINTENANCE": "MAINT"
})

print("\nAfter Normalization:")
print(df["Status"].unique())

print("\nStatus Counts:")
print(df["Status"].value_counts())
```

```python
print("Before Validation:")
print(df[["Stack_ID", "Plant_ID"]].head())

df["Mapping_Status"] = "Valid"

invalid_stack = df.groupby("Stack_ID")["Plant_ID"].transform("nunique") > 1

df.loc[invalid_stack, "Mapping_Status"] = "Invalid"

print("\nAfter Validation:")
print(df[["Stack_ID", "Plant_ID", "Mapping_Status"]].head())

print("\nMapping Status Counts:")
print(df["Mapping_Status"].value_counts())
```

```python
print("Units Before Conversion:")
print(df["Unit"].value_counts())

mask = df["Unit"] == "mg/Nm3"

df.loc[mask, "PM2_5"] = df.loc[mask, "PM2_5"] * 1000
df.loc[mask, "SO2"] = df.loc[mask, "SO2"] * 1000
df.loc[mask, "NOx"] = df.loc[mask, "NOx"] * 1000

df.loc[mask, "Unit"] = "ug/m3"

print("\nUnits After Conversion:")
print(df["Unit"].value_counts())

print("\nSample Data:")
print(df[["PM2_5", "SO2", "NOx", "Unit"]].head())
```

```python
print("Source Types:")
print(df["Source_Type"].value_counts())

df["Source_Tag"] = df["Source_Type"].replace({
    "ambient": "Ambient Monitoring",
    "stack": "Stack Monitoring",
    "lab": "Other",
    "inspection": "Other"
})

print("\nSource Tags:")
print(df["Source_Tag"].value_counts())

print("\nSample Data:")
print(df[["Source_Type", "Source_Tag"]].head())
```

```python
print("Double Entry Check:")

print(df["Double_entry_check_flag"].value_counts())

df["QC_Status"] = df["Double_entry_check_flag"].replace({
    1: "Passed",
    0: "Failed"
})

print("\nQC Status:")
print(df["QC_Status"].value_counts())

print("\nSample Data:")
print(df[["Lab_sample_ID", "Double_entry_check_flag", "QC_Status"]].head())
```
